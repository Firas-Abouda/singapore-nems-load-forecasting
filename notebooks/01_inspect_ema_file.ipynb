```python
monday = raw.iloc[5:53, 0:4].copy()

monday.columns = [
    "Period Ending Time",
    *raw.iloc[4, 1:4].tolist(),
]

date = pd.Timestamp(raw.iloc[1, 1]).normalize()

period_ending = monday.pop("Period Ending Time").astype(str)

timestamp = pd.to_datetime(
    date.strftime("%Y-%m-%d") + " " + period_ending,
    format="%Y-%m-%d %H:%M",
    errors="raise",
)

timestamp.loc[period_ending.eq("00:00")] += pd.Timedelta(days=1)

monday.insert(
    0,
    "Timestamp",
    timestamp.dt.tz_localize("Asia/Singapore"),
)

monday = monday.reset_index(drop=True)

monday
```

```python
assert len(monday) == 48
assert monday["Timestamp"].is_unique
assert monday["Timestamp"].is_monotonic_increasing
```

```python
columns = raw.iloc[4, 1:4].tolist()

values = raw.iloc[5:53, 1:22].to_numpy()

# 48 × 7 × 3
# → 7 × 48 × 3
# → 336 × 3
values = (
    values
    .reshape(48, 7, 3)
    .transpose(1, 0, 2)
    .reshape(336, 3)
)


dates = pd.DatetimeIndex(
    pd.to_datetime(raw.iloc[1, 1:22:3])
)

expected_dates = pd.date_range(
    start=dates[0],
    periods=7,
    freq="D",
)

if not dates.equals(expected_dates):
    raise ValueError("Unexpected dates in EMA weekly file")

timestamps = pd.date_range(
    start=dates[0] + pd.Timedelta(minutes=30),
    periods=336,
    freq="30min",
    tz="Asia/Singapore",
)

week = pd.DataFrame(values, columns=columns)
week.insert(0, "Timestamp", timestamps)



week
```

```python
assert len(week) == 336
assert week["Timestamp"].is_unique
assert week["Timestamp"].is_monotonic_increasing
```
